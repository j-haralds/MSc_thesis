```python
import numpy as np
import pandas as pd
from tqdm import trange
import matplotlib.pyplot as plt

import sys
sys.path.append('..')
import plot_settings
plot_settings.apply()
```

```python
def sigmoid(x,a):
    return 1/(1+np.exp(-a*x))

def ReLU(x):
    return np.maximum(0,x)

def softplus(x):
    return np.log(1 + np.exp(x))

f,ax = plt.subplots(1,4,figsize=(10,2.75))
x = np.linspace(-10,10,100)
for i in range(4):
    ax[i].set_xlabel(r"$s$")
    ax[i].grid()
ax[0].set_ylabel(r"$A(s)$")

ax[0].plot(x,sigmoid(x,1), lw = 3)
ax[0].set_title("Sigmoid")
ax[2].plot(x,ReLU(x), lw = 3)
ax[2].set_title(r"ReLU")
# ax[2].plot(x,softmax(x), lw = 2)
# ax[2].set_title("Softmax")
ax[3].plot(x,softplus(x), lw = 3)
ax[3].set_title("Softplus")
ax[1].plot(x,np.tanh(x), lw = 3)
ax[1].set_title("Tanh")



plt.tight_layout()
plt.savefig('report_figs/activation_funcs.pdf',bbox_inches='tight')

plt.show()
```
